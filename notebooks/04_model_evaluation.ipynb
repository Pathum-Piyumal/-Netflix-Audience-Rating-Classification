# Phase 4: Model Evaluation & Performance Analysis
This notebook covers:
1. Loading the trained model artifact (`models/best_rating_model.joblib`)
2. Test Set Evaluation Metrics (Accuracy, Precision, Recall, Macro/Weighted F1)
3. Confusion Matrix Plotting & Analysis
4. Class-wise Performance Breakdown

```python
import os
import sys
sys.path.append('..')

import joblib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
from src.preprocessing import clean_data, get_train_test_data
from src.evaluate import evaluate_and_save_results
```

## 1. Load Trained Model & Test Set Split

```python
model = joblib.load('../models/best_rating_model.joblib')
df_raw = pd.read_csv('../data/raw/netflix_titles.csv')
df_clean = clean_data(df_raw)
_, X_test, _, y_test = get_train_test_data(df_clean)

print(f"Test set sample size: {len(X_test)}")
```

## 2. Evaluate Performance & Render Confusion Matrix

```python
y_pred = model.predict(X_test)
print("Classification Report:\n")
print(classification_report(y_test, y_pred))

# Render Confusion Matrix Heatmap
labels = sorted(list(y_test.unique()))
cm = confusion_matrix(y_test, y_pred, labels=labels)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title('Rating Classification Confusion Matrix')
plt.xlabel('Predicted Rating Category')
plt.ylabel('True Rating Category')
plt.tight_layout()
plt.show()
```
