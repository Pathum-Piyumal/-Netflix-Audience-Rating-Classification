# Phase 1: Data Exploration & Understanding
This notebook covers:
1. Dataset Loading & Inspection
2. Missing Values Analysis
3. Target Variable (`rating`) Distribution & Grouping
4. Exploratory Data Analysis (EDA) on Movies vs TV Shows, Genres, and Release Years

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot aesthetics
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Load Dataset & Basic Inspection

```python
# Load raw dataset
df = pd.read_csv('../data/raw/netflix_titles.csv')

print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
# Dataset Summary Statistics & Data Types
df.info()
```

## 2. Missing Values Analysis

```python
# Count and percentage of missing values per column
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing Percentage (%)': missing_pct})
missing_df[missing_df['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False)
```

## 3. Target Analysis: Content Ratings

```python
# Target Variable Value Counts
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='rating', order=df['rating'].value_counts().index, palette='viridis')
plt.title('Distribution of Content Ratings on Netflix')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../reports/figures/rating_distribution.png')
plt.show()
```

## 4. EDA: Content Type vs Rating

```python
# Movies vs TV Shows Ratings breakdown
plt.figure(figsize=(14, 6))
sns.countplot(data=df, x='rating', hue='type', order=df['rating'].value_counts().index, palette='Set2')
plt.title('Content Rating Distribution by Type (Movie vs TV Show)')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../reports/figures/rating_by_type.png')
plt.show()
```
